```python
def visualize_diffusion(steps, vis_steps):
    """
    steps: Tensor [T, B, 1, H, W]
    vis_steps: list of timesteps to visualize
    """

    T, B, _, H, W = steps.shape

    fig, axes = plt.subplots(
        nrows=B,
        ncols=len(vis_steps),
        figsize=(len(vis_steps) * 2, B * 2),
    )

    for row in range(B):
        for col, t in enumerate(vis_steps):
            img = steps[t, row, 0].cpu()

            axes[row, col].imshow(img, cmap="gray")
            axes[row, col].axis("off")

            if row == 0:
                axes[row, col].set_title(f"t={t+1}")

    plt.tight_layout()
    plt.savefig("figs")
    plt.show()
```

```python
visualize_diffusion(steps, vis_steps)
```

```python
import matplotlib.pyplot as plt
import numpy as np

def visualize_diffusion(steps, vis_steps):
    """
    steps: Tensor [T, B, 1, H, W]
    vis_steps: list of timesteps
    """

    T, B, _, H, W = steps.shape

    fig, axes = plt.subplots(
        nrows=B,
        ncols=len(vis_steps),
        figsize=(len(vis_steps) * 2, B * 2),
    )

    # 🔥 Make axes always 2D
    axes = np.atleast_2d(axes)

    for row in range(B):
        for col, t in enumerate(vis_steps):
            img = steps[t, row, 0].cpu()

            # normalize if needed
            # img = (img + 1) / 2

            axes[row, col].imshow(img, cmap="gray")
            axes[row, col].axis("off")

            if row == 0:
                axes[row, col].set_title(f"t={t}")

    plt.tight_layout()
    plt.show()
```
