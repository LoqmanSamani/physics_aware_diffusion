```python
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
from diffusion.schedules import LinearVS
from diffusion.forward import ForwardVP
from diffusion.reverse import ReverseVP
```

```python
data = np.zeros((9, 100, 100))
data[:, 30:70, 30:70] = 1.0
```

```python
plt.imshow(data[8, :, :])
```

```python
beta_start = 0.3
beta_end = 5.0
start = 0.0
end = 1.0
num_steps = 1000
t = torch.linspace(start, end, num_steps)
print(t.shape)
```

```python
betas = beta_start + (beta_end - beta_start) * t / end
```

```python
#betas
```

```python
beta_integral = beta_start * t + 0.5 * (beta_end - beta_start) * t ** 2 / end
```

```python
alpha = torch.exp(-0.5 * beta_integral)
#print(alpha)
```

```python
sigma = torch.sqrt(1.0 - alpha**2)
#print(sigma)
```

```python
alpha_2 = alpha ** 2
```

```python
#print(alpha_2)
```

```python
x0 = torch.from_numpy(data)
#print(x0)
print(x0.shape)
```

```python
noise = torch.randn_like(x0)#, dtype=torch.float32)
#print(noise)
```

```python
xt = alpha * x0 + sigma * noise
```

```python
fig = plt.figure(figsize=(8, 8))
axis = fig.subplots(3, 3)
axis[0, 0].imshow(xt[:, :, 1])
axis[0, 1].imshow(xt[:, :, 200]) 
axis[0, 2].imshow(xt[:, :, 300])
axis[1, 0].imshow(xt[:, :, 400])
axis[1, 1].imshow(xt[:, :, 500])
axis[1, 2].imshow(xt[:, :, 600])
axis[2, 0].imshow(xt[:, :, 700])
axis[2, 1].imshow(xt[:, :, 800])
axis[2, 2].imshow(xt[:, :, -1])
plt.show()
```

```python
vs = LinearVS(num_steps, beta_start, beta_end, start, end)
```

```python
vp = ForwardVP(vs)
```

```python
noise = torch.randn_like(x0)
xT = vp(x0, noise, torch.full((9,), 999))
```

```python
fig = plt.figure(figsize=(8, 8))
axis = fig.subplots(3, 3)
axis[0, 0].imshow(xT[0, :, :])
axis[0, 1].imshow(xT[1, :, :]) 
axis[0, 2].imshow(xT[2, :, :])
axis[1, 0].imshow(xT[3, :, :])
axis[1, 1].imshow(xT[4, :, :])
axis[1, 2].imshow(xT[5, :, :])
axis[2, 0].imshow(xT[6, :, :])
axis[2, 1].imshow(xT[7, :, :])
axis[2, 2].imshow(xT[8, :, :])
plt.show()
```

```python
rvp = ReverseVP(vs)
```

```python
x_current = xT.clone()

for step in range(999, 0, -1):
    t = torch.full((9,), step)

    alpha_t = torch.exp(-0.5 * rvp.vs.integral_beta[t])
    sigma_t = rvp.vs.get_std(t)

    while alpha_t.dim() < x_current.dim():
        alpha_t = alpha_t.unsqueeze(-1)
        sigma_t = sigma_t.unsqueeze(-1)

    perfect_noise = (x_current - alpha_t * x0) / sigma_t
    x_current = rvp(x_current, perfect_noise, t, noise)
```

```python
plt.imshow(x_current[8])
plt.title("Recovered x0")
plt.show()
```
